```python
from tensorboard.backend.event_processing.event_accumulator import EventAccumulator
import glob
import os

event_files = glob.glob(
    "train_logs/Peract_push_buttons_ndf/**/events.out.tfevents*",
    recursive=True
)

for file in event_files:
    print("\nFILE:", file)

    ea = EventAccumulator(os.path.dirname(file))
    ea.Reload()

    print("Scalars:", ea.Tags()["scalars"])
    print("Images:", ea.Tags()["images"])
    print("Histograms:", ea.Tags()["histograms"])
```

```python
from tensorboard.backend.event_processing.event_accumulator import EventAccumulator

logdir = "train_logs/Peract_push_buttons_ndf/denoise3d-PeractTwoCam-C120-B16-lr1e-4-constant-H3-rectified_flow"

ea = EventAccumulator(logdir)
ea.Reload()

for tag in ea.Tags()["scalars"]:
    events = ea.Scalars(tag)
    print(
        tag,
        "Number of points:", len(events),
        "First:", events[0].value if events else None,
        "Last:", events[-1].value if events else None
    )
```
